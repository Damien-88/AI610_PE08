### Students will develop a Python program where:
#### You are a data scientist for a real estate agency, and your task is to build a model that predicts house prices based on multiple features:
####	1.	Size (sq ft): Area of the house.
####	2.	Number of Bedrooms: Count of bedrooms in the house.
####	3.	Age (years): Age of the house.

#### You will train a multivariate linear regression model on the provided dataset and evaluate its performance.
#### * Students must include meaningful comments in their code that explain key logic and decisions.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Dataset: features (size, bedrooms, age) and prices
features = np.array([[1500, 3, 20], [1700, 3, 15], [2000, 4, 10], [2100, 4, 8], [2500, 5, 5]])
prices = np.array([300000, 340000, 400000, 420000, 500000])

# Train the Linear Regression model
model = LinearRegression()
model.fit(features, prices)

# Predict price for a new house
new_house = np.array([[2200, 4, 7]])
predicted_price = model.predict(new_house)

# Model evaluation: Mean Squared Error
predicted_prices = model.predict(features)
mse = mean_squared_error(prices, predicted_prices)

# Calculate model coefficients, intercept, and R-squared value
coefficients = model.coef_
intercept = model.intercept_
r2 = model.score(features, prices)

# Output predictions and evaluation
print(f"Predicted price for the new house: ${predicted_price[0]:,.2f}")
print(f"Model Mean Squared Error: {mse:.2f}")
print(f"Model Coefficients: {coefficients}")
print(f"Model Intercept: {intercept}")
print(f"Model R-squared Value: {r2:.2f}")
```
